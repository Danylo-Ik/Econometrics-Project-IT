```python
df2020 = df2020.rename(columns={'Ви працюєте зараз в ІТ?': 'full_time', 
                                'Де ви мешкаєте?': 'region',
                                'Де ви працюєте?': 'work_site',
                                'Вкажіть заклад вищої освіти': 'university',
                                'Оберіть вашу посаду': 'position',
                                'Ваш тайтл': 'title',
                                'В якій сфері працюєте?': 'industry',
                                'Тип компанії': 'company_type',
                                'Стаж на поточному місці роботи': 'tenure',
                                'Вкажіть суму цього бонуса у $$$ після податків': 'bonus',
                                'Фреймворки, бібліотеки та платформи': 'frameworks',})

df2020['full_time'] = df2020.apply(lambda x: 1 if x['full_time'] == 'Так, працюю full-time' else 0, axis=1)

df2020['programming_language'] = df2020['programming_language'].fillna(df2020['Основна мова програмування'])

df2020['title'] = df2020['title'].fillna(df2020['Ваш тайтл_1']).fillna(df2020['Ваш тайтл_4']).fillna(df2020['Ваш тайтл_5']).fillna(df2020['Ваш тайтл_7']).fillna(df2020['Ваш тайтл_9']).fillna(df2020['Ваш тайтл_11']).fillna(df2020['Ваш тайтл_13']).fillna(df2020['Ваш тайтл_15'])

df2020.drop(columns=['Вкажіть вашу спеціалізацію', 'Вкажіть вашу спеціалізацію_2', 'Інші мови програмування', 'Ваш тайтл_1',
                     'Ваш тайтл_4', 'Ваш тайтл_5', 'Ваш тайтл_7', 'Ваш тайтл_9', 'Ваш тайтл_11', 'Ваш тайтл_13', 'Ваш тайтл_15',
                     'Оберіть вашу посаду_3', 'Ваша посада', 'Оберіть вашу посаду_6', 'Ваша посада_8', 'Ваша посада_10',
                     'Ваша посада_12', 'Ваша посада_14', 'Ваша посада_16', 'Чи отримуєте ви грошові бонуси до зарплати?', 'Основна мова програмування',
                     "У скількох компаніях встигли попрацювати за час ІТ-кар'єри?", 'Платформи, для яких розробляєте на поточному місці роботи?'], inplace=True)
```

```python
def convert_tenure(value):
    if "рік" in value or "роки" in value or "років" in value:
        return float(value.split()[0])
    elif "10 і більше" in value:
        return 10
    elif "Менше як 3 місяці" in value:
        return 0.25
    elif "Пів року" in value:
        return 0.5
    elif "3 місяці" in value:
        return 0.25
    return None  # For unexpected values

df2020['tenure'] = df2020['tenure'].apply(convert_tenure)
```

### Salary change

```python
print('salary_change',df2020['salary_change'].unique())
print("\n")
print('salary_change',df2021['salary_change'].unique())
print("\n")
```

```python
def extract_numeric(val):
    if isinstance(val, str):
        val = val.replace(',', '.').replace('$', '').replace('%', '').strip()
        if val.startswith('+'):
            val = val[1:]
        try:
            return float(val)
        except:
            return None
    return val

df2020['salary_change'] = df2020['salary_change'].apply(extract_numeric)
df2021['salary_change'] = df2021['salary_change'].apply(extract_numeric)
```
